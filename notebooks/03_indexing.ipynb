```
import json
from services.indexing_service import (
    build_inverted_index, get_doc_lengths,
    get_avg_doc_length, save_index, load_index, get_index_stats
)

# Load processed docs
print('Loading CT2021 docs...')
with open(f'{SAVE_DIR}/ct2021_docs_processed.json') as f:
    docs1 = json.load(f)
print(f'CT2021: {len(docs1):,} docs loaded')

print('Loading MSMARCO docs...')
with open(f'{SAVE_DIR}/msmarco_docs_processed.json') as f:
    docs2 = json.load(f)
print(f'MSMARCO: {len(docs2):,} docs loaded')
```

```
# Build index for Dataset 1 - Clinical Trials
print('Building inverted index for CT2021...')
index1 = build_inverted_index(docs1)
doc_lengths1 = get_doc_lengths(docs1)

stats1 = get_index_stats(index1, doc_lengths1)
print('\n=== CT2021 Index Stats ===')
for k, v in stats1.items():
    print(f'  {k}: {v:,}' if isinstance(v, int) else f'  {k}: {v}')

save_index(index1, f'{SAVE_DIR}/ct2021_index.pkl')

# save doc lengths too
with open(f'{SAVE_DIR}/ct2021_doc_lengths.json', 'w') as f:
    json.dump(doc_lengths1, f)
print('CT2021 indexing done')
```

```
# Build index for Dataset 2 - MSMARCO
print('Building inverted index for MSMARCO...')
index2 = build_inverted_index(docs2)
doc_lengths2 = get_doc_lengths(docs2)

stats2 = get_index_stats(index2, doc_lengths2)
print('\n=== MSMARCO Index Stats ===')
for k, v in stats2.items():
    print(f'  {k}: {v:,}' if isinstance(v, int) else f'  {k}: {v}')

save_index(index2, f'{SAVE_DIR}/msmarco_index.pkl')

with open(f'{SAVE_DIR}/msmarco_doc_lengths.json', 'w') as f:
    json.dump(doc_lengths2, f)
print('MSMARCO indexing done')
```

```
# Verify: test a term lookup
test_term = 'diabetes'

if test_term in index1:
    docs_with_term = list(index1[test_term].items())[:5]
    print(f'CT2021 — term "{test_term}" found in {len(index1[test_term]):,} docs')
    print('  Sample:', docs_with_term)
else:
    print(f'Term "{test_term}" not in CT2021 index')

if test_term in index2:
    print(f'MSMARCO — term "{test_term}" found in {len(index2[test_term]):,} docs')
else:
    print(f'Term "{test_term}" not in MSMARCO index')

print('\n=== Indexing Complete ===')
print('Next: 04_retrieval_models.ipynb')
```
